# Baseline Model v1 — Training Results

Traffic Sign Classifier (GTSRB) — first unregularized CNN, 10 epochs.

```python
import matplotlib.pyplot as plt

epochs = list(range(1, 11))
train_loss = [1.6183, 0.3635, 0.2021, 0.1351, 0.1028, 0.0708, 0.0611, 0.0537, 0.0475, 0.0377]
test_acc = [71.51, 78.38, 80.78, 81.72, 84.62, 85.64, 85.10, 82.18, 85.34, 84.88]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(epochs, train_loss, marker='o')
axes[0].set_title('Training Loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')

axes[1].plot(epochs, test_acc, marker='o', color='orange')
axes[1].set_title('Test Accuracy')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)')

plt.tight_layout()
plt.show()
```

## Interpretation

*We can see a gap between the loss and the accuracy as it dips in epoch 8 then plateaus yet the loss keeps going lower which tells us that its trying to memorize . epoch 6 was the peak  accuracy(85.34%) and the model only saves epoch 10 regardless , so we end up keeping a slightly worse then epoch 6*
