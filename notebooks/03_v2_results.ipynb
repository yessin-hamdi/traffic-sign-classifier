# v2 Results — Augmentation + Dropout + Early Stopping

Traffic Sign Classifier (GTSRB) — CNN with data augmentation (rotation, translation, color jitter), dropout (p=0.5) on the FC layer, and early-stopping checkpointing (patience=5).

```python
import matplotlib.pyplot as plt

epochs_v2 = list(range(1, 31))
train_loss_v2 = [2.7440, 1.8532, 1.5402, 1.3239, 1.1951, 1.0785, 0.9893, 0.9323, 0.8739, 0.8285,
                  0.7796, 0.7440, 0.7078, 0.6871, 0.6504, 0.6207, 0.5967, 0.5761, 0.5548, 0.5338,
                  0.5132, 0.5077, 0.4873, 0.4750, 0.4647, 0.4549, 0.4462, 0.4357, 0.4204, 0.4158]
test_acc_v2 = [42.93, 56.65, 61.92, 66.95, 70.35, 74.08, 74.64, 77.78, 79.49, 80.92,
               80.86, 82.95, 83.58, 84.40, 84.66, 87.24, 86.56, 86.78, 86.86, 86.81,
               87.28, 87.36, 87.85, 87.66, 88.58, 88.73, 89.45, 89.46, 89.63, 89.37]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(epochs_v2, train_loss_v2, marker='o', markersize=3)
axes[0].set_title('v2 Training Loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')

axes[1].plot(epochs_v2, test_acc_v2, marker='o', markersize=3, color='orange')
axes[1].set_title('v2 Test Accuracy')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy (%)')

plt.tight_layout()
plt.show()
```

## v1 vs v2 comparison

```python
epochs_v1 = list(range(1, 11))
test_acc_v1 = [71.51, 78.38, 80.78, 81.72, 84.62, 85.64, 85.10, 82.18, 85.34, 84.88]

plt.figure(figsize=(9, 5))
plt.plot(epochs_v1, test_acc_v1, marker='o', label='v1 (baseline, no regularization)')
plt.plot(epochs_v2, test_acc_v2, marker='o', label='v2 (augmentation + dropout)')
plt.axhline(y=85.64, color='gray', linestyle='--', linewidth=1, label='v1 peak (85.64%)')
plt.axhline(y=89.63, color='green', linestyle='--', linewidth=1, label='v2 peak (89.63%)')
plt.title('Test Accuracy: Baseline v1 vs v2')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()
plt.tight_layout()
plt.show()
```
